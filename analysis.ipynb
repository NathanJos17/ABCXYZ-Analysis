```python
import pandas as pd
import numpy as np
import glob
```

```python
filepath = glob.glob('sales-regional-*.parquet')

# Parquet keeps the source schema and avoids CSV mixed-type warnings.
df = pd.concat([pd.read_parquet(file) for file in filepath], ignore_index=True)
df["ARTICLE"] = df["ARTICLE"].astype("string")
df = df.drop(columns=[col for col in df.columns if col.startswith('GR_')], errors='ignore')
```

```python
df["price"] = df["ARTICLE"].map(masterart.set_index("Article")["Loc Amount"])
df["ProdClass"] = df["ARTICLE"].map(masterart.set_index("Article")["Product Classification"])
df["lv3"] = df["ARTICLE"].map(masterart.set_index("Article")["Lv3-Description"])
df["lv4"] = df["ARTICLE"].map(masterart.set_index("Article")["Lv4-Description"])
df["Article Description"] = df["ARTICLE"].map(masterart.set_index("Article")["Article Description"])
df["generic"] = df["ARTICLE"].map(masterart.set_index("Article")["Generic Code"])
df["Article Type"] = df["ARTICLE"].map(masterart.set_index("Article")["Article Type"])
```

```python
import gc

for temporary_name in ['sales_long', 'monthly_article', 'sales_by_article', 'article_summary', 'monthly_groups']:
    if temporary_name in globals():
        del globals()[temporary_name]
gc.collect()

sales_cols = [
    col for col in df.columns
    if col.startswith('QTY_SALES_') and col.endswith('_2025')
]
group_cols = ['ARTICLE','generic','SITE_ID', 'ProdClass', 'lv3', 'lv4']

sales_by_article = df.groupby(
    group_cols,
    sort=False,
    observed=True,
    dropna=False
)[sales_cols].sum().reset_index()

site_names = df[group_cols + ['SITE_NAME']].drop_duplicates(group_cols)
prices = df[['ARTICLE', 'price']].drop_duplicates('ARTICLE')
sales_by_article = sales_by_article.merge(site_names, on=group_cols, how='left')
sales_by_article = sales_by_article.merge(prices, on='ARTICLE', how='left')

sales_by_article[sales_cols] = sales_by_article[sales_cols].apply(
    pd.to_numeric,
    errors='coerce'
).fillna(0)
sales_by_article['price'] = pd.to_numeric(
    sales_by_article['price'],
    errors='coerce'
).fillna(0)

monthly_qty = sales_by_article[sales_cols]
sales_by_article['total_qty'] = monthly_qty.sum(axis=1)
sales_by_article['avg_monthly_qty'] = monthly_qty.mean(axis=1)
sales_by_article['total_value'] = sales_by_article['price'] * sales_by_article['total_qty']
sales_by_article['avg_monthly_value'] = sales_by_article['price'] * sales_by_article['avg_monthly_qty']
sales_by_article['cv'] = monthly_qty.std(axis=1, ddof=0).div(
    monthly_qty.mean(axis=1).replace(0, np.nan)
).fillna(0)

article_summary = sales_by_article[
    ['ARTICLE', 'generic','SITE_ID', 'SITE_NAME', 'ProdClass', 'lv3', 'lv4', 'price',
     'total_qty', 'avg_monthly_qty', 'total_value', 'avg_monthly_value', 'cv']
].copy()

ranking_cols = ['SITE_ID', 'ProdClass', 'lv3', 'lv4']
article_summary = article_summary.sort_values(
    ranking_cols + ['total_value'],
    ascending=[True, True, True, True, False]
).reset_index(drop=True)

article_summary['cum_value'] = article_summary.groupby(
    ranking_cols,
    observed=True,
    dropna=False
)['total_value'].cumsum()
article_summary['group_total_value'] = article_summary.groupby(
    ranking_cols,
    observed=True,
    dropna=False
)['total_value'].transform('sum')
article_summary['cum_pct'] = np.where(
    article_summary['group_total_value'] != 0,
    article_summary['cum_value'] / article_summary['group_total_value'],
    0
)

article_summary['abc_class'] = np.select(
    [article_summary['cum_pct'] <= 0.80, article_summary['cum_pct'] <= 0.95],
    ['A', 'B'],
    default='C'
)

article_summary['xyz_class'] = np.select(
    [
        article_summary['total_qty'] == 0,
        article_summary['cv'] < 0.5,
        (article_summary['cv'] > 0.5) & (article_summary['cv'] < 1)
    ],
    ['N', 'X', 'Y'],
    default='Z'
)

article_summary['abc_xyz'] = article_summary['abc_class'] + article_summary['xyz_class']

article_summary.head(50)
```

```python
from openpyxl import Workbook
from openpyxl.styles import Alignment, Border, Font, PatternFill, Side
from openpyxl.utils import get_column_letter

abcxyz_order = ['AX', 'AY', 'AZ', 'AN', 'BX', 'BY', 'BZ', 'BN', 'CX', 'CY', 'CZ', 'CN']
site_order = sorted(article_summary['SITE_ID'].dropna().unique())

sum_sales = article_summary.pivot_table(
    index='abc_xyz',
    columns='SITE_ID',
    values='total_value',
    aggfunc='sum',
    fill_value=0
)
count_article = article_summary.pivot_table(
    index='abc_xyz',
    columns='SITE_ID',
    values='ARTICLE',
    aggfunc='nunique',
    fill_value=0
)

abcxyz_site_summary = pd.concat(
    {'Sum Sales': sum_sales, 'Count Article': count_article},
    axis=1
).swaplevel(0, 1, axis=1)
abcxyz_site_summary = abcxyz_site_summary.reindex(
    index=abcxyz_order,
    columns=pd.MultiIndex.from_product(
        [site_order, ['Sum Sales', 'Count Article']],
        names=['SITE ID', None]
    ),
    fill_value=0
)
abcxyz_site_summary.index.name = 'ABCXYZ RESULT'

csv_path = 'abcxyz_site_summary.csv'
excel_path = 'abcxyz_site_summary.xlsx'
abcxyz_site_summary.to_csv(csv_path)

workbook = Workbook()
worksheet = workbook.active
worksheet.title = 'ABCXYZ Summary'
worksheet['A1'] = 'SITE ID'
worksheet['A2'] = 'ABCXYZ RESULT'

header_fill = PatternFill('solid', fgColor='D9EAF7')
thin_side = Side(style='thin', color='B7B7B7')
cell_border = Border(left=thin_side, right=thin_side, top=thin_side, bottom=thin_side)

for site_index, site_id in enumerate(site_order):
    first_column = 2 + site_index * 2
    last_column = first_column + 1
    worksheet.merge_cells(
        start_row=1,
        start_column=first_column,
        end_row=1,
        end_column=last_column
    )
    worksheet.cell(1, first_column).value = site_id
    worksheet.cell(2, first_column).value = 'Sum Sales'
    worksheet.cell(2, last_column).value = 'Count Article'

for row_index, abcxyz_class in enumerate(abcxyz_order, start=3):
    worksheet.cell(row_index, 1).value = abcxyz_class
    for site_index, site_id in enumerate(site_order):
        first_column = 2 + site_index * 2
        worksheet.cell(row_index, first_column).value = abcxyz_site_summary.loc[
            abcxyz_class, (site_id, 'Sum Sales')
        ]
        worksheet.cell(row_index, first_column + 1).value = abcxyz_site_summary.loc[
            abcxyz_class, (site_id, 'Count Article')
        ]

for row in worksheet.iter_rows(
    min_row=1,
    max_row=worksheet.max_row,
    min_col=1,
    max_col=worksheet.max_column
):
    for cell in row:
        cell.border = cell_border
        cell.alignment = Alignment(horizontal='center', vertical='center')
        if cell.row <= 2:
            cell.font = Font(bold=True)
            cell.fill = header_fill
        elif cell.column > 1:
            cell.number_format = '#,##0'

worksheet.column_dimensions['A'].width = 18
for column_index in range(2, worksheet.max_column + 1):
    worksheet.column_dimensions[get_column_letter(column_index)].width = 16

worksheet.freeze_panes = 'B3'
workbook.save(excel_path)

abcxyz_site_summary.style.format('{:,.0f}', na_rep='0')
```

```python
from openpyxl import Workbook
from openpyxl.styles import Alignment, Border, Font, PatternFill, Side
from openpyxl.utils import get_column_letter

abcxyz_order = ['AX', 'AY', 'AZ', 'AN', 'BX', 'BY', 'BZ', 'BN', 'CX', 'CY', 'CZ', 'CN']
excel_path = 'abcxyz_summary_fixed.xlsx'

summary_by_class = article_summary.groupby('abc_xyz', as_index=False).agg(
    count_article=('ARTICLE', 'nunique'),
    total_qty=('total_qty', 'sum'),
    total_value=('total_value', 'sum')
)
summary_by_class['abc_xyz'] = pd.Categorical(
    summary_by_class['abc_xyz'],
    categories=abcxyz_order,
    ordered=True
)
summary_by_class = summary_by_class.sort_values('abc_xyz').reset_index(drop=True)

abc_counts = article_summary.groupby('abc_class', as_index=False).agg(
    count_article=('ARTICLE', 'nunique'),
    sales_qty=('total_qty', 'sum'),
    sales_value=('total_value', 'sum')
).set_index('abc_class').reindex(['A', 'B', 'C'], fill_value=0).reset_index()

xyz_counts = article_summary.groupby('xyz_class', as_index=False).agg(
    count_article=('ARTICLE', 'nunique'),
    sales_qty=('total_qty', 'sum'),
    sales_value=('total_value', 'sum')
).set_index('xyz_class').reindex(['N', 'X', 'Y', 'Z'], fill_value=0).reset_index()

matrix_groupby = article_summary.groupby(
    ['abc_class', 'xyz_class'],
    observed=True
).agg(
    count_generic=('generic', 'nunique'),
    sales_qty=('total_qty', 'sum'),
    sales_value=('total_value', 'sum')
)

abcxyz_generic_matrix = matrix_groupby['count_generic'].unstack(fill_value=0).reindex(
    index=['A', 'B', 'C'],
    columns=['N', 'X', 'Y', 'Z'],
    fill_value=0
)
abcxyz_qty_matrix = matrix_groupby['sales_qty'].unstack(fill_value=0).reindex(
    index=['A', 'B', 'C'],
    columns=['N', 'X', 'Y', 'Z'],
    fill_value=0
)
abcxyz_value_matrix = matrix_groupby['sales_value'].unstack(fill_value=0).reindex(
    index=['A', 'B', 'C'],
    columns=['N', 'X', 'Y', 'Z'],
    fill_value=0
)

summary_by_site = article_summary.pivot_table(
    index='abc_xyz',
    columns='SITE_ID',
    values='total_value',
    aggfunc='sum',
    fill_value=0
).reindex(abcxyz_order, fill_value=0)

workbook = Workbook()
workbook.remove(workbook.active)

sheets = {
    'ABCXYZ Summary': summary_by_class,
    'Sales by Site': summary_by_site,
    'ABC Class Counts': abc_counts,
    'XYZ Class Counts': xyz_counts
}

header_fill = PatternFill('solid', fgColor='D9EAF7')
thin_side = Side(style='thin', color='B7B7B7')
cell_border = Border(left=thin_side, right=thin_side, top=thin_side, bottom=thin_side)

def style_sheet(worksheet):
    for row in worksheet.iter_rows():
        for cell in row:
            cell.border = cell_border
            cell.alignment = Alignment(horizontal='center', vertical='center')
            if cell.row == 1:
                cell.font = Font(bold=True)
                cell.fill = header_fill
            elif cell.column > 1:
                cell.number_format = '#,##0'
    for column_index in range(1, worksheet.max_column + 1):
        worksheet.column_dimensions[get_column_letter(column_index)].width = 18
    worksheet.freeze_panes = 'A2'

for sheet_name, dataframe in sheets.items():
    worksheet = workbook.create_sheet(sheet_name)
    output_dataframe = dataframe.reset_index() if dataframe.index.name else dataframe
    worksheet.append(list(output_dataframe.columns))
    for row in output_dataframe.itertuples(index=False, name=None):
        worksheet.append(list(row))
    style_sheet(worksheet)

matrix_sheet = workbook.create_sheet('ABCXYZ Matrix')
matrix_tables = [
    ('Sales Qty Matrix', abcxyz_qty_matrix),
    ('Sales Value Matrix', abcxyz_value_matrix),
    ('Count Generic Matrix', abcxyz_generic_matrix)
]

for table_index, (table_title, matrix) in enumerate(matrix_tables):
    if table_index > 0:
        matrix_sheet.append([])
    matrix_sheet.append([table_title])
    matrix_sheet.append(['ABC Class'] + list(matrix.columns))
    for row_index, row in matrix.iterrows():
        matrix_sheet.append([row_index] + list(row))

style_sheet(matrix_sheet)
matrix_sheet.freeze_panes = 'B3'

workbook.save(excel_path)
print(f'Saved ABCXYZ summary to {excel_path}')
```

```python
from openpyxl import Workbook
from openpyxl.styles import Alignment, Border, Font, PatternFill, Side
from openpyxl.utils import get_column_letter

abcxyz_order = ['AX', 'AY', 'AZ', 'AN', 'BX', 'BY', 'BZ', 'BN', 'CX', 'CY', 'CZ', 'CN']
excel_path = 'abcxyz_site_summary_with_names.xlsx'

# Create SITE_ID -> SITE_NAME mapping
site_mapping = (
    article_summary[['SITE_ID', 'SITE_NAME']]
    .dropna(subset=['SITE_ID'])
    .drop_duplicates('SITE_ID')
    .set_index('SITE_ID')['SITE_NAME']
    .to_dict()
)

site_order = sorted(
    article_summary['SITE_ID'].dropna().unique(),
    key=lambda value: str(value)
)

site_labels = {}
for site_id in site_order:
    site_id_text = str(site_id).replace('.0', '')
    site_name = site_mapping.get(site_id, '')
    site_labels[site_id] = (
        f'{site_id_text} - {site_name}'
        if site_name
        else site_id_text
    )

# Table 1: Count Generic
count_generic = article_summary.pivot_table(
    index='abc_xyz',
    columns='SITE_ID',
    values='generic',
    aggfunc='nunique',
    fill_value=0
).reindex(
    index=abcxyz_order,
    columns=site_order,
    fill_value=0
)

# Table 2: Sales Value
sales_value = article_summary.pivot_table(
    index='abc_xyz',
    columns='SITE_ID',
    values='total_value',
    aggfunc='sum',
    fill_value=0
).reindex(
    index=abcxyz_order,
    columns=site_order,
    fill_value=0
)

workbook = Workbook()
worksheet = workbook.active
worksheet.title = 'ABCXYZ Summary'

header_fill = PatternFill('solid', fgColor='D9EAF7')
title_fill = PatternFill('solid', fgColor='B4C7E7')
thin_side = Side(style='thin', color='B7B7B7')
cell_border = Border(
    left=thin_side,
    right=thin_side,
    top=thin_side,
    bottom=thin_side
)

def write_table(start_row, title, dataframe, number_format):
    total_columns = len(site_order) + 1

    # Table title
    worksheet.merge_cells(
        start_row=start_row,
        start_column=1,
        end_row=start_row,
        end_column=total_columns
    )
    title_cell = worksheet.cell(start_row, 1)
    title_cell.value = title
    title_cell.font = Font(bold=True)
    title_cell.fill = title_fill
    title_cell.alignment = Alignment(horizontal='center')

    # Header row
    header_row = start_row + 1
    worksheet.cell(header_row, 1).value = 'ABCXYZ RESULT'

    for column_index, site_id in enumerate(site_order, start=2):
        worksheet.cell(
            header_row,
            column_index
        ).value = site_labels[site_id]

    # Data rows
    for row_index, abcxyz_class in enumerate(abcxyz_order, start=header_row + 1):
        worksheet.cell(row_index, 1).value = abcxyz_class

        for column_index, site_id in enumerate(site_order, start=2):
            worksheet.cell(
                row_index,
                column_index
            ).value = dataframe.loc[abcxyz_class, site_id]

    # Styling
    for row in worksheet.iter_rows(
        min_row=start_row,
        max_row=header_row + len(abcxyz_order),
        min_col=1,
        max_col=total_columns
    ):
        for cell in row:
            cell.border = cell_border
            cell.alignment = Alignment(
                horizontal='center',
                vertical='center'
            )

            if cell.row in [start_row, header_row]:
                cell.font = Font(bold=True)
                cell.fill = header_fill
            elif cell.column > 1:
                cell.number_format = number_format

    worksheet.row_dimensions[header_row].height = 32

    return header_row + len(abcxyz_order) + 2


next_row = write_table(
    start_row=1,
    title='Count Generic by Site',
    dataframe=count_generic,
    number_format='#,##0'
)

write_table(
    start_row=next_row,
    title='Sales Value by Site',
    dataframe=sales_value,
    number_format='#,##0'
)

worksheet.column_dimensions['A'].width = 18

for column_index in range(2, worksheet.max_column + 1):
    worksheet.column_dimensions[
        get_column_letter(column_index)
    ].width = 22

worksheet.freeze_panes = 'B3'

workbook.save(excel_path)
print(f'Saved ABCXYZ summary to {excel_path}')
```
